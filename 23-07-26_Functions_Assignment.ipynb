```python
#Using reduce(), calculate the sum of unique numbers only. Do not directly use sum(set(numbers)).
from functools import reduce

numbers = [12, 45, 12, 67, 45, 89, 23, 67, 100]

unique_sum = reduce(lambda total, x: total + x if x not in numbers[:numbers.index(x)] else total, numbers, 0)

print(unique_sum)
```

```python
#Section 5: Combined Lambda, map(), filter(),
# and reduce() — 20 Questions

'''Using filter() and map():

Extract all even numbers.
Calculate the square of every even number.
Return the squared values as a list.'''

numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

squared_even_numbers = list(
    map(lambda x: x ** 2, filter(lambda x: x % 2 == 0, numbers))
)

print(squared_even_numbers)
```

```python
'''Using filter() and reduce():

Extract all odd numbers.
Calculate their total sum.'''

from functools import reduce

numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

sum = reduce(
    lambda a, b: a + b,
    filter(lambda x: x % 2 != 0, numbers)
)

print(sum)
```

```python
'''Using filter() and reduce():

Extract palindrome words.
Join them into one string separated by " | ".'''

from functools import reduce

words = ["madam", "python", "level", "azure", "radar", "cloud", "civic"]

result = reduce(
    lambda a, b: a + " | " + b,
    filter(lambda word: word == word[::-1], words)
)

print(result)
```

```python
'''Using filter(), map(), and reduce():

Extract temperatures greater than 30°C.
Convert the selected values into Fahrenheit.
Calculate the average Fahrenheit temperature'''

from functools import reduce

temperatures = [18, 24, 30, 35, 40, 12, 28, 33, 15, 42]

# Extract temperatures > 30°C, convert to Fahrenheit, and calculate average
fahrenheit_avg = reduce(
    lambda a, b: a + b,
    map(lambda c: (c * 9/5) + 32, filter(lambda c: c > 30, temperatures))
) / len(list(filter(lambda c: c > 30, temperatures)))

print(fahrenheit_avg)
```

```python
'''Using filter() and reduce():

Extract students who passed.
Find the topper among the passed students.'''

from functools import reduce

students = [
    {"name": "Ravi", "marks": 92, "passed": True},
    {"name": "Kiran", "marks": 68, "passed": True},
    {"name": "Meena", "marks": 85, "passed": True},
    {"name": "Ajay", "marks": 48, "passed": True},
    {"name": "Sneha", "marks": 35, "passed": False},
    {"name": "Rahul", "marks": 95, "passed": True}
]

topper = reduce(
    lambda s1, s2: s1 if s1["marks"] > s2["marks"] else s2,
    filter(lambda student: student["marks"] >= 40, students)
)

print(topper)
```

```python
'''Using filter(), map(), and reduce():

Extract students scoring more than 70.
Extract only their marks.
Calculate the average marks.'''

from functools import reduce

students = [
    {"name": "Ravi", "marks": 92},
    {"name": "Kiran", "marks": 68},
    {"name": "Meena", "marks": 85},
    {"name": "Ajay", "marks": 48},
    {"name": "Sneha", "marks": 79},
    {"name": "Rahul", "marks": 95}
]

passed_students = filter(lambda student: student["marks"] > 70, students)

marks = list(map(lambda student: student["marks"], passed_students))

average_marks = reduce(lambda a, b: a + b, marks) / len(marks)

print(average_marks)
```

```python
'''Using filter(), map(), and reduce():

Extract products currently in stock.
Calculate each product's inventory value using price × stock.
Calculate the total inventory value.'''

from functools import reduce

products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]

total_inventory_value = reduce(
    lambda total, value: total + value,
    map(
        lambda product: product["price"] * product["stock"],
        filter(lambda product: product["stock"], products)
    ),
    0
)

print(total_inventory_value)
```

```python
'''Using filter(), map(), and reduce():

Extract active salespeople from the South region.
Calculate a 5% incentive for each selected salesperson.
Create a list containing salesperson name, sales amount, and incentive.
Calculate the total incentive payable.'''




from functools import reduce



sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]

# Extract active salespeople from South region and calculate 5% incentive
sales_list = list(map(
    lambda person: {
        "Salesperson Name": person["salesperson"],
        "Sales Amount": person["sales"],
        "Incentive": person["sales"] * 0.05
    },
    filter(lambda person: person["region"] == "South" and person["active"], sales_records)
))

# Calculate total incentive payable
total_incentive = reduce(
    lambda total, person: total + person["Incentive"],
    sales_list,
    0
)

print(sales_list)
print("Total Incentive Payable:", total_incentive)
```
