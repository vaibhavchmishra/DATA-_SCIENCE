```python
#Check whether a number is finite,infinite,or Nan using math.isfinite(),math.isinf(),and math.isnan().
import math

# Define different types of numbers
numbers = [100, float('inf'), float('nan')]

for num in numbers:
    if math.isfinite(num):
        print(num, "is finite")
    elif math.isinf(num):
        print(num, "is infinite")
    elif math.isnan(num):
        print(num, "is NaN")
```

```python
#Find the sum of square roots of all numbers in a list.

import math

# Define a list of numbers
numbers = [4, 9, 16, 25]

# Find the sum of the square roots
result = sum(math.sqrt(num) for num in numbers)

# Display the result
print("Sum of square roots =", result)
```

```python
#Calculate compound interest using math.pow()

import math

# Define the values
principal = 1000       # Principal amount
rate = 8                # Annual interest rate in percent
time = 2                # Time in years

# Calculate compound interest
amount = principal * math.pow(1 + rate / 100, time)
compound_interest = amount - principal

# Display the results
print("Compound Interest =", compound_interest)
print("Total Amount =", amount)
```

```python
#Find the area and circumference of a circle using math.pi.

import math

# Define the radius of the circle
radius = 5

# Calculate the area
area = math.pi * math.pow(radius, 2)

# Calculate the circumference
circumference = 2 * math.pi * radius

# Display the results
print("Area of the circle =", area)
print("Circumference of the circle =", circumference)
```

```python
#Create a scientifc calculator that support square root, factorial ,logarithm,power,trigonometric functions,GCD, and LCM.


import math

def scientific_calculator():
    while True:
        print("\n===== SCIENTIFIC CALCULATOR =====")
        print("1. Square Root")
        print("2. Factorial")
        print("3. Logarithm (base 10)")
        print("4. Power")
        print("5. Sine")
        print("6. Cosine")
        print("7. Tangent")
        print("8. GCD")
        print("9. LCM")
        print("10. Exit")

        choice = int(input("Enter your choice: "))

        if choice == 1:
            num = float(input("Enter a number: "))
            print("Square Root =", math.sqrt(num))

        elif choice == 2:
            num = int(input("Enter a non-negative integer: "))
            print("Factorial =", math.factorial(num))

        elif choice == 3:
            num = float(input("Enter a positive number: "))
            print("Logarithm (base 10) =", math.log10(num))

        elif choice == 4:
            base = float(input("Enter the base: "))
            exponent = float(input("Enter the exponent: "))
            print("Result =", math.pow(base, exponent))

        elif choice == 5:
            angle = float(input("Enter angle in degrees: "))
            print("Sine =", math.sin(math.radians(angle)))

        elif choice == 6:
            angle = float(input("Enter angle in degrees: "))
            print("Cosine =", math.cos(math.radians(angle)))

        elif choice == 7:
            angle = float(input("Enter angle in degrees: "))
            print("Tangent =", math.tan(math.radians(angle)))

        elif choice == 8:
            num1 = int(input("Enter first integer: "))
            num2 = int(input("Enter second integer: "))
            print("GCD =", math.gcd(num1, num2))

        elif choice == 9:
            num1 = int(input("Enter first integer: "))
            num2 = int(input("Enter second integer: "))
            print("LCM =", math.lcm(num1, num2))

        elif choice == 10:
            print("Exiting calculator...")
            break

        else:
            print("Invalid choice! Please try again.")


scientific_calculator()
```
